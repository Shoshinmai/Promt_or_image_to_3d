```python
from rembg import remove
from PIL import Image
from io import BytesIO
import requests
import os
```

```python
url = "https://static.vecteezy.com/system/resources/previews/030/688/265/large_2x/horse-2d-vector-illustration-cartoon-in-white-background-h-free-photo.jpg"
image_name = url.split("/")[-1]
image_name
```

```python
# img = Image.open(r"D:\AI_dev\ML\project\prompt_or_image_to_3d_model\original\ce3d8eb9e258bc0b1c904b28d9823291.jpg")
img = Image.open(BytesIO(requests.get(url).content))
img.save("original/"+image_name, format='jpeg')
img
```

```python
out_path = 'masked/'+image_name
out_path
```

```python
with open(out_path, 'wb') as f:
    inp = open('original/'+image_name, 'rb').read()
    sub = remove(inp)
    f.write(sub)
```
